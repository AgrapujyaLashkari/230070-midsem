## Task 1.2: Key Assumptions

**Assumption 1**
* **Assumption:** The semantic relationships among categories in the tree are strictly generalization-specialization.
* **Why the method needs it:** The orthogonal transfer regularization ($\frac{1}{2} \sum K_{ij} |w_i^T w_j|$) attempts to strip away general properties learned by ancestors so child nodes focus on distinct properties.
* **Violation scenario:** A taxonomy where children are entirely different concepts grouped arbitrarily, rather than specific instances of the parent.
* **Paper reference:** Section 1, Paragraph 4.

**Assumption 2**
* **Assumption:** The comparison matrix $\overline{K}$ derived from the parameter matrix $K$ is positive semidefinite.
* **Why the method needs it:** This is a strict mathematical requirement to ensure the proposed optimization problem remains convex, guaranteeing convergence to a global minimum.
* **Violation scenario:** Selecting arbitrary penalty weights for the regularization term that result in a matrix with negative eigenvalues.
* **Paper reference:** Section 3.1, Theorem 1.

**Assumption 3**
* **Assumption:** Classification tasks are easier at higher levels and become progressively more difficult toward the bottom of the tree.
* **Why the method needs it:** The method relies on this assumption to justify assigning larger classification margins to higher-level classifications (by setting larger values in the diagonal matrix $K_{ii}$).
* **Violation scenario:** A dataset where distinguishing between the top-level categories is actually harder due to noisy data than distinguishing fine-grained leaf nodes.
* **Paper reference:** Section 3.1, Equation (6) and following paragraphs.
