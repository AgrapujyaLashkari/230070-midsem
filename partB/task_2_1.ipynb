## Task 2.1: Dataset Selection and Setup
* **What the dataset is**: I selected the scikit-learn Wine dataset. It is a well-known public toy dataset with 178 samples and 13 features across 3 classes, making it a perfect realistic substitute to test a 3-class hierarchical tree without the perfect separability of purely synthetic blobs.
* **Why it is a reasonable testbed**: It allows for the construction of a small, controlled 3-class hierarchy (Root -> Node 1 & Node 2 -> Node 3 & Node 4) to test the core orthogonal transfer mathematical formulation without the computational overhead of NLP TF-IDF vectorization.
* **Limitations**: It lacks the natural, semantic generalization-specialization text structure present in the original RCV1-v2 dataset used by the authors.

```python
import pandas as pd
import os
from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler

os.makedirs('data', exist_ok=True)

# Load the real-world 'Wine' toy dataset
wine = load_wine()
X = wine.data
y = wine.target

# Standardize features (crucial for SVMs on real data)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Save to CSV
df = pd.DataFrame(X_scaled, columns=wine.feature_names)
df['target'] = y
df.to_csv('data/wine_dataset.csv', index=False)
print("Wine dataset generated and saved to partB/data/wine_dataset.csv")
```
