## Task 1.3: What the Paper Claims to Improve

* **Main Baseline**: The paper primarily compares against TREELOSS (Tree-induced loss SVM).
* **Limitation of Baseline**: Tree-induced loss functions downplay the difficulty of classifying similar sibling classes at lower levels by simply assigning them smaller penalties based on short graph distances.
* **Improvement**: Instead of reducing the penalty, the proposed method tackles the classification difficulty directly by forcing the model to learn new, orthogonal features at the lower levels rather than reusing the parent's features.
* **Condition where it would not outperform**: The method would likely fail to outperform the baseline if the lower-level classes actually rely on the exact same feature combinations as their parents, making orthogonal hyperplanes detrimental to accuracy.
